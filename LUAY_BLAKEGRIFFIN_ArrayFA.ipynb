```python
scores = [80, 85, 90, 95, 100]

# Function 1: Prints the current list of student scores
def display_scores(scores_list):
  print("\nStudent Scores:")
  print(scores_list)

# Function 2: Searches for a specific score entered by the user
def search_score(scores_list):
  # Ask the user for a score and convert the input to an integer
  search = int(input("Enter score to search: "))

  # Check if the entered score exists in the list
  if search in scores_list:
    print("Score found!")
  else:
    print("Score not found.")

# Function 3: Calculates and displays the average of all scores
def compute_average(scores_list):
  # Divide the total sum of scores by the number of items in the list
  average = sum(scores_list) / len(scores_list)
  print("Average score:", average)

# Function 4: Updates a score at a specific 0-based index (0 to 4)
def update_score(scores_list):
  print("Current Scores:", scores_list)

  # Ask the user which index position they want to modify
  position = int(input("Enter index number (0-4): "))

  # Validate that the index is within the valid list range (0 to 4)
  if 0 <= position < len(scores_list):
    new_score = int(input("Enter new score: "))
    # Replace the old score at that index with the new score
    scores_list[position] = new_score
    print("Score successfully updated!")
  else:
    print("Invalid index number.")

# Function 5: Displays the list after modifications have been made
def display_updated_scores(scores_list):
  print("\nUpdated Student Scores:")
  print(scores_list)


# Main function that controls the menu and calls the other functions
def main():
  # Keep showing the menu repeatedly until the user chooses to exit
  while True:
    print("\n=== Student Score Analyzer ===")
    print("1. Display Scores")
    print("2. Search Score")
    print("3. Compute Average Score")
    print("4. Update Score")
    print("5. Display Updated Scores")
    print("6. Exit")

    # Get the user's menu selection
    choice = input("\nEnter your choice: ")

    # Match the user's choice to the corresponding function call
    if choice == "1":
      display_scores(scores)
    elif choice == "2":
      search_score(scores)
    elif choice == "3":
      compute_average(scores)
    elif choice == "4":
      update_score(scores)
    elif choice == "5":
      display_updated_scores(scores)
    elif choice == "6":
      print("Thank you for using Student Score Analyzer!")
      break
    else:
      # Handle any input that isn't 1 through 6
      print("Invalid choice. Please try again.")

# Call the main function to start running the program
main()
```
